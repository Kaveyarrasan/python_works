```python
# Global dictionary to simulate a database
account_holder_dict = {}

class BankAccountSystem:
    def __init__(self, acc_no):
        self.account_number = acc_no
        self.__balance = account_holder_dict.get(acc_no, {}).get('balance', 0)
        account_holder_dict[self.account_number] = {'balance': self.__balance}

    # Deposit method
    def deposit(self, amount):
        if amount <= 0:
            print("Deposit Transaction Declined")
            print("Reason: Amount must be positive")
            return False

        print("Started Deposit Transaction")
        self.__balance += amount
        account_holder_dict[self.account_number]['balance'] = self.__balance
        print(f"Deposited amount {amount} added to account: {self.account_number}")
        return True

    # Withdraw method
    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal Transaction Declined")
            print("Reason: Amount must be positive")
            return False

        if amount > self.__balance:
            print("Withdrawal Transaction Declined")
            print("Reason: Insufficient balance")
            return False

        print("Started Withdrawal Transaction")
        self.__balance -= amount
        account_holder_dict[self.account_number]['balance'] = self.__balance
        print(f"Withdrawn amount {amount} from account: {self.account_number}")
        return True

    # Balance check method
    def check_balance(self):
        msg_dict = {
            "Account Holder": self.account_number,
            "Balance": self.__balance
        }
        msg = f"Account Holder: {self.account_number}\nBalance Amount: {self.__balance}"
        print(msg)
        return msg_dict

    # Getter for balance (read-only)
    def get_balance(self):
        return self.__balance


# Example usage
acc_holder = BankAccountSystem('ACCXNOIS00001')
acc_holder.check_balance()

acc_holder.deposit(500)
acc_holder.check_balance()

acc_holder.withdraw(200)
acc_holder.check_balance()

acc_holder.withdraw(400)  # Should decline due to insufficient balance
```
