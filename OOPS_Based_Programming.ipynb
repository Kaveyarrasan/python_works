```python
account_holder_dict = {}   # Declaring empty dictionary data type to capture Account Holders Data
```

```python
"""
Bank Account System

Create a BankAccount class with attributes: account_number, balance.

Methods: deposit(amount), withdraw(amount), check_balance().

Problem: Implement deposit/withdraw logic with encapsulation (balance should not be directly modified outside the class).
"""

class BankAccountSystem:
    def __init__(self, acc_no):
        self.account_number = acc_no
        self.__balance = account_holder_dict.get(acc_no, {}).get('balance', 0)
        account_holder_dict[self.account_number] = {'balance': self.__balance}

        # if self.account_number not in account_holder_dict:
        #     # print("Initializing BankAccount System")
        #     self.balance = 0
        #     account_holder_dict[self.account_number] = {'balance': self.balance}
        # else:
        #     self.balance = account_holder_dict[self.account_number]['balance']

    def deposit(self, amount):
        try:
            if self.account_number in account_holder_dict:
                print("Started Deposit Transaction")
                self.__balance += amount
                account_holder_dict[self.account_number].update({'balance': self.__balance})
                print(f"Deposited amount {amount} added to account: {self.account_number}")
                self.check_balance()
                return True
                # return check_balance()
            else:
                print("Deposit Transaction Declined")
                print(f"Reason: Account: {self.account_number} not exists in database {account_holder_dict}")
                return False
        except Exception as e:
            print(e)
            return False

    def withdraw(self, amount):
        try:
            if self.account_number in account_holder_dict:
                if amount <= account_holder_dict[self.account_number]['balance']:
                    print("Started Withdrawal Transaction")
                    self.__balance -= amount
                    account_holder_dict[self.account_number].update({'balance': self.__balance})
                    print(f"Withdrawn amount {amount} from account: {self.account_number}")
                    self.check_balance()
                    return True
                else:
                    print("Withdrawal Transaction Declined")
                    print(f"Reason: Unable to withdraw amount {amount} as insufficient balance {self.__balance} in account: {self.account_number}")
                    return False
                # return check_balance()
            else:
                print("Withdrawal Transaction Declined")
                print(f"Reason: Account: {self.account_number} not exists in database {account_holder_dict}")
                return False
        except Exception as e:
            print(e)
            return False

    def check_balance(self):
        try:
            msg_dict = {
                "Account Holder": self.account_number,
                "Balance": account_holder_dict[self.account_number]['balance']
            }
            msg = f"Account Holder: {self.account_number}\nBalance Amount: {account_holder_dict[self.account_number]['balance']}"
            print(msg)
            # return msg_dict
        except Exception as e:
            print(e)
            return False
```
